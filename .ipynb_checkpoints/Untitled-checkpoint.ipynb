```python
# Q1: Record a speech, plot it, and save the signal in Excel

import sounddevice as sd
import scipy.io.wavfile as wav
import matplotlib.pyplot as plt
import pandas as pd

fs = 8000
duration = 2

print("Start speaking...")
recording = sd.rec(int(duration * fs), samplerate=fs, channels=1)
sd.wait()
print("Recording complete")

wav.write("speech.wav", fs, recording)

fs, signal = wav.read("speech.wav")

plt.plot(signal)
plt.xlabel("Samples")
plt.ylabel("Amplitude")
plt.title("Speech Signal")
plt.show()

df = pd.DataFrame({"Speech Signal": signal.flatten()})
df.to_excel("question1.xlsx", index=False)

print("Saved successfully!")
```

```python
# Q2: Sample the speech signal at 256 Hz and compare the samples

import scipy.io.wavfile as wav
import matplotlib.pyplot as plt

fs, signal = wav.read("speech.wav")
signal = signal.flatten()

original_samples = len(signal)

new_fs = 256
step = fs // new_fs

sampled_signal = signal[::step]

sampled_samples = len(sampled_signal)

print("Original sampling frequency:", fs, "Hz")
print("New sampling frequency:", new_fs, "Hz")
print("Original number of samples:", original_samples)
print("After sampling:", sampled_samples)

plt.plot(sampled_signal)
plt.xlabel("Samples")
plt.ylabel("Amplitude")
plt.title("Speech Signal Sampled at 256 Hz")
plt.show()
```

```python
# Q3: Divide speech signal into 1-second segments

import scipy.io.wavfile as wav
import os

fs, signal = wav.read("speech.wav")

segment_size = fs       # 1 second
num_segments = len(signal) // segment_size

os.makedirs("segments", exist_ok=True)

for i in range(num_segments):
    start = i * segment_size
    end = start + segment_size
    
    segment = signal[start:end]
    
    wav.write(f"segments/segment_{i+1}.wav", fs, segment)

print("Number of segments:", num_segments)
print("Segments saved successfully!")
```

```python
# Q4: Find Fourier Transform of a signal and check periodicity

import numpy as np
import matplotlib.pyplot as plt

# Create a signal
fs = 8000
t = np.arange(0, 2, 1/fs)

signal = np.sin(2 * np.pi * 100 * t)

# Fourier Transform
fft_signal = np.fft.fft(signal)
freq = np.fft.fftfreq(len(signal), 1/fs)

# Magnitude spectrum
magnitude = np.abs(fft_signal)

# Plot
plt.plot(freq[:len(freq)//2], magnitude[:len(magnitude)//2])
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.title("Fourier Transform of Signal")
plt.show()

# Find dominant frequency
peak = np.argmax(magnitude[:len(magnitude)//2])
print("Dominant frequency:", freq[peak], "Hz")

if magnitude[peak] > 0:
    print("Periodic component detected.")
```

```python
# Q5: DFT of a sine wave

import numpy as np
import matplotlib.pyplot as plt

fs = 8000
t = np.arange(0, 1, 1/fs)

signal = np.sin(2 * np.pi * 100 * t)

dft = np.fft.fft(signal)
freq = np.fft.fftfreq(len(signal), 1/fs)

# Plot only 0-500 Hz
plt.plot(freq[:len(freq)//2], np.abs(dft[:len(dft)//2]))

plt.xlim(0, 500)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.title("DFT of Sine Wave")
plt.grid()

plt.show()
```

```python
# Q6: Find FFT of a sine wave using Python

import numpy as np
import matplotlib.pyplot as plt

fs = 8000
t = np.arange(0, 1, 1/fs)

signal = np.sin(2 * np.pi * 100 * t)

# FFT
fft_signal = np.fft.fft(signal)
freq = np.fft.fftfreq(len(signal), 1/fs)

# Plot only 0-500 Hz
plt.plot(freq[:len(freq)//2], np.abs(fft_signal[:len(fft_signal)//2]))

plt.xlim(0, 500)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.title("FFT of Sine Wave")
plt.grid()

plt.show()
```

```python
# Q9: Find FFT and STFT coefficients and plot spectrogram

import numpy as np
import scipy.io.wavfile as wav
import scipy.signal as signal
import matplotlib.pyplot as plt

# Read speech recording
fs, speech = wav.read("speech.wav")
speech = speech.flatten()

# FFT
fft_signal = np.fft.fft(speech)
freq = np.fft.fftfreq(len(speech), 1/fs)

# Plot FFT
plt.plot(freq[:len(freq)//2], np.abs(fft_signal[:len(fft_signal)//2]))
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.title("FFT of Speech Signal")
plt.xlim(0, 4000)
plt.grid()
plt.show()

# STFT
f, t, Zxx = signal.stft(speech, fs=8000)

# Spectrogram
plt.pcolormesh(t, f, np.abs(Zxx), shading='gouraud')
plt.xlabel("Time (seconds)")
plt.ylabel("Frequency (Hz)")
plt.title("Spectrogram of Speech Signal")
plt.colorbar(label="Magnitude")
plt.show()
```

```python
# Q10/11: Find Mel Spectrogram of speech recording

import numpy as np
import librosa
import librosa.display
import matplotlib.pyplot as plt

# Read speech recording
speech, fs = librosa.load("speech.wav", sr=8000)

# Calculate Mel Spectrogram
mel = librosa.feature.melspectrogram(
    y=speech,
    sr=fs,
    n_mels=40
)

# Convert to decibels
mel_db = librosa.power_to_db(mel, ref=np.max)

# Plot Mel Spectrogram
librosa.display.specshow(
    mel_db,
    sr=fs,
    x_axis="time",
    y_axis="mel"
)

plt.colorbar(label="dB")
plt.title("Mel Spectrogram of Speech")
plt.show()
```

```python
# Q12: Find MFCC features from speech recording and plot them

import librosa
import librosa.display
import matplotlib.pyplot as plt

# Read speech recording
speech, fs = librosa.load("speech.wav", sr=8000)

# Calculate MFCC
mfcc = librosa.feature.mfcc(
    y=speech,
    sr=fs,
    n_mfcc=13
)

# Plot MFCC
plt.figure(figsize=(10, 5))
librosa.display.specshow(
    mfcc,
    sr=fs,
    x_axis="time"
)

plt.colorbar()
plt.xlabel("Time (seconds)")
plt.ylabel("MFCC Coefficients")
plt.title("MFCC Features of Speech")
plt.show()
```
