```python
# Q1: Study the libraries required for video processing in Python.
# Record a video and store it in any format.
# Convert the video format to MP4 from MOV or AVI.

import cv2

cap = cv2.VideoCapture("input_video.mov")

if not cap.isOpened():
    print("Error: Cannot open input_video.mov")
else:
    fps = cap.get(cv2.CAP_PROP_FPS)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    fourcc = cv2.VideoWriter_fourcc(*"mp4v")
    out = cv2.VideoWriter(
        "output_video.mp4", fourcc, fps, (width, height)
    )

    while True:
        ret, frame = cap.read()
        if not ret:
            break
        out.write(frame)

    cap.release()
    out.release()
    print("Video converted to output_video.mp4")
```

```python
# Q2: Convert the video to frames and count the number
# of frames produced.

import cv2
import os

cap = cv2.VideoCapture("input_video.mov")
os.makedirs("video_frames", exist_ok=True)

count = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    cv2.imwrite(f"video_frames/frame_{count+1}.jpg", frame)
    count += 1

cap.release()

print("Total frames:", count)
print("Frames saved in video_frames folder")
```
